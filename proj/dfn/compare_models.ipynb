```python
import pybamm
import numpy as np
import pickle
import matplotlib.pyplot as plt

pybamm.set_logging_level("INFO")

data_DIR = "../data/"
fig_DIR = "../figs/"
```

```python
with open(data_DIR +'spm_old_fresh_vars.pickle', 'rb') as handle:
        spm_old = pickle.load(handle)
with open(data_DIR +'dfn_old_fresh_vars.pickle', 'rb') as handle:
        dfn_old = pickle.load(handle)
with open(data_DIR +'spm_new_fresh_vars.pickle', 'rb') as handle:
        spm_new = pickle.load(handle)
with open(data_DIR +'dfn_new_fresh_vars.pickle', 'rb') as handle:
        dfn_new = pickle.load(handle)
```

```python
fig, ax = plt.subplots(2,1,figsize=(5,8))
ax1 = ax.flat[0]
ax1.plot(spm_old[0],spm_old[1],'b')
ax1.plot(spm_new[0],spm_new[1],'r--')
ax1.set_xlabel("Time [s]")
ax1.set_ylabel("Voltage [V]")
ax2 = ax.flat[1]
ax2.plot(spm_old[0],spm_old[2],'b')
ax2.plot(spm_new[0],spm_new[2],'r--')
ax2.set_xlabel("Time [s]")
ax2.set_ylabel("Current [A]")
ax1.legend(["Old","New"])
fig.suptitle("1C SPM Old vs New")
fig.savefig(fig_DIR+"spm_comp_1C.png",dpi=600)
```

```python
fig, ax = plt.subplots(2,1,figsize=(5,8))
ax1 = ax.flat[0]
ax1.plot(dfn_old[0],dfn_old[1],'b')
ax1.plot(dfn_new[0],dfn_new[1],'r--')
ax1.set_xlabel("Time [s]")
ax1.set_ylabel("Voltage [V]")
ax2 = ax.flat[1]
ax2.plot(dfn_old[0],dfn_old[2],'b')
ax2.plot(dfn_new[0],dfn_new[2],'r--')
ax2.set_xlabel("Time [s]")
ax2.set_ylabel("Current [A]")
ax1.legend(["Old","New"])
fig.suptitle("1C DFN Old vs New")
fig.savefig(fig_DIR+"dfn_comp_1C.png",dpi=600)
```
