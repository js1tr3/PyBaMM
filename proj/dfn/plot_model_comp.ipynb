```python
import pybamm
import numpy as np
import pickle
import matplotlib.pyplot as plt

pybamm.set_logging_level("INFO")

data_DIR = "../data/"
fig_DIR = "../figs/"
```

## Charge Discharge 1 Cycle

```python
spm_old = {}
with open(data_DIR +'spm_old_fresh_vars_Cb5.pickle', 'rb') as handle:
        spm_old[0] = pickle.load(handle)
with open(data_DIR +'spm_old_fresh_vars_1p5C.pickle', 'rb') as handle:
        spm_old[1] = pickle.load(handle)
with open(data_DIR +'spm_old_fresh_vars_2C.pickle', 'rb') as handle:
        spm_old[2] = pickle.load(handle)
spm_new = {}
with open(data_DIR +'spm_new_fresh_vars_Cb5.pickle', 'rb') as handle:
        spm_new[0] = pickle.load(handle)
with open(data_DIR +'spm_new_fresh_vars_1p5C.pickle', 'rb') as handle:
        spm_new[1] = pickle.load(handle)
with open(data_DIR +'spm_new_fresh_vars_2C.pickle', 'rb') as handle:
        spm_new[2] = pickle.load(handle)
dfn_old = {}
with open(data_DIR +'dfn_old_fresh_vars_Cb5.pickle', 'rb') as handle:
        dfn_old[0] = pickle.load(handle)
with open(data_DIR +'dfn_old_fresh_vars_1p5C.pickle', 'rb') as handle:
        dfn_old[1] = pickle.load(handle)
with open(data_DIR +'dfn_old_fresh_vars_2C.pickle', 'rb') as handle:
        dfn_old[2] = pickle.load(handle)
dfn_new = {}
with open(data_DIR +'dfn_new_fresh_vars_Cb5.pickle', 'rb') as handle:
        dfn_new[0] = pickle.load(handle)
with open(data_DIR +'dfn_new_fresh_vars_1p5C.pickle', 'rb') as handle:
        dfn_new[1] = pickle.load(handle)
with open(data_DIR +'dfn_new_fresh_vars_2C.pickle', 'rb') as handle:
        dfn_new[2] = pickle.load(handle)
```

```python
fig, ax = plt.subplots(2,3,figsize=(15,8))
ax1 = ax.flat[0]
ax1.plot(spm_old[0][1],spm_old[0][2],'b')
ax1.plot(spm_new[0][1],spm_new[0][2],'r--')
ax1.set_xlabel("Time [h]")
ax1.set_ylabel("Voltage [V]")
ax1.set_title("C/5")
ax1.legend(["Old","New"])
ax2 = ax.flat[3]
ax2.plot(spm_old[0][1],spm_old[0][3],'b')
ax2.plot(spm_new[0][1],spm_new[0][3],'r--')
ax2.set_xlabel("Time [h]")
ax2.set_ylabel("Current [A]")

ax3 = ax.flat[1]
ax3.plot(spm_old[1][1],spm_old[1][2],'b')
ax3.plot(spm_new[1][1],spm_new[1][2],'r--')
ax3.set_xlabel("Time [h]")
ax3.set_ylabel("Voltage [V]")
ax3.set_title("1.5C")
ax4 = ax.flat[4]
ax4.plot(spm_old[1][1],spm_old[1][3],'b')
ax4.plot(spm_new[1][1],spm_new[1][3],'r--')
ax4.set_xlabel("Time [h]")
ax4.set_ylabel("Current [A]")

ax5 = ax.flat[2]
ax5.plot(spm_old[2][1],spm_old[2][2],'b')
ax5.plot(spm_new[2][1],spm_new[2][2],'r--')
ax5.set_xlabel("Time [h]")
ax5.set_ylabel("Voltage [V]")
ax5.set_title("2C")
ax6 = ax.flat[5]
ax6.plot(spm_old[2][1],spm_old[2][3],'b')
ax6.plot(spm_new[2][1],spm_new[2][3],'r--')
ax6.set_xlabel("Time [h]")
ax6.set_ylabel("Current [A]")

fig.suptitle("SPM Old vs New")
fig.tight_layout()
fig.savefig(fig_DIR+"spm_comp_multi.png",dpi=600)
```

```python
with open(data_DIR +'spm_old_fresh_vars_pulse.pickle', 'rb') as handle:
        spm_old = pickle.load(handle)
with open(data_DIR +'spm_new_fresh_vars_pulse.pickle', 'rb') as handle:
        spm_new = pickle.load(handle)
with open(data_DIR +'dfn_old_fresh_vars_pulse.pickle', 'rb') as handle:
        dfn_old = pickle.load(handle)
with open(data_DIR +'dfn_new_fresh_vars_pulse.pickle', 'rb') as handle:
        dfn_new = pickle.load(handle)
```

```python
fig, ax = plt.subplots(2,3,figsize=(15,8))
ax1 = ax.flat[0]
ax1.plot(spm_old[0],spm_old[12],'b')
ax1.plot(spm_new[0],spm_new[12],'r--')
ax1.set_xlim([50,100])
ax1.set_xlabel("Time [s]")
ax1.set_ylabel("Voltage [V]")
ax1.set_title("Reaction overpotential")
ax1.legend(["Old","New"])
ax2 = ax.flat[1]
ax2.plot(spm_old[0],spm_old[13],'b')
ax2.plot(spm_new[0],spm_new[13],'r--')
ax2.set_xlim([50,100])
ax2.set_xlabel("Time [s]")
ax2.set_ylabel("Voltage [V]")
ax2.set_title("Concentration overpotential")
ax3 = ax.flat[2]
ax3.plot(spm_old[0],spm_old[14],'b')
ax3.plot(spm_new[0],spm_new[14],'r--')
ax3.set_xlim([50,100])
ax3.set_xlabel("Time [s]")
ax3.set_ylabel("Voltage [V]")
ax3.set_title("Ohmic electrolyte overpotential")
ax4 = ax.flat[3]
ax4.plot(spm_old[0],spm_old[15],'b')
ax4.plot(spm_new[0],spm_new[15],'r--')
ax4.set_xlim([50,100])
ax4.set_xlabel("Time [s]")
ax4.set_ylabel("Voltage [V]")
ax4.set_title("Ohmic electrode overpotential")
ax5 = ax.flat[5]
ax5.plot(spm_old[0],spm_old[16],'b')
ax5.plot(spm_new[0],spm_new[16],'r--')
ax5.set_xlim([50,100])
ax5.set_xlabel("Time [s]")
ax5.set_ylabel("Voltage [V]")
ax5.set_title("Battery open circuit voltage [V]")
fig.suptitle("SPM: Pulse at 50% SOC")
fig.tight_layout()
fig.savefig(fig_DIR+"spm_comp_pulse.png",dpi=600)
```
