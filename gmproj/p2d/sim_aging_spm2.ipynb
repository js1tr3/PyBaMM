```python
def get_c(sols):
    Cn = []
    for i in range(len(sols)):
        sol_long = sols[i]
        Cn1 = sol_long.all_summary_variables
        Cn2 = Cn1[0]['C']
        Cn.append(Cn2)
    Cn = np.array(Cn)
    return Cn
```
